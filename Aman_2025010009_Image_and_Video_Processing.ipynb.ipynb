```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("car.jpg")

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("IMAGE")
plt.axis("off")
plt.show()

cv2.imwrite("car_copy.jpg", img)
print("Image saved successfully.")
```

```python
import cv2

img = cv2.imread("car.jpg")

cv2.imwrite("car.png", img)
cv2.imwrite("car.tiff", img)

print("Image converted successfully.")
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("car.jpg")
resized = cv2.resize(img, (400, 300))

plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized Image")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("car.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
import cv2

video = cv2.VideoCapture("video.mp4")

frame_no = 0

while True:
    ret, frame = video.read()

    if not ret:
        break

    frame_no += 1

video.release()

print("Video read successfully.")
print("Total Frames:", frame_no)
```

```python
import cv2

video = cv2.VideoCapture("video.mp4")

count = 0

while True:
    ret, frame = video.read()

    if not ret:
        break

    cv2.imwrite(f"frame_{count}.jpg", frame)
    count += 1

video.release()

print("Frames extracted successfully.")
print("Total Frames Extracted:", count)
```
